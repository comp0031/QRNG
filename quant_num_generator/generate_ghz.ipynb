```python
from dotenv import load_dotenv
import os

from tripartite_diqrng.ghz_generate import generate, get_bounds

load_dotenv()
```

# Simulated Backend

```python
uncertainty = 0.5
BASES = [("X", "X", "X"), ("X", "Y", "Y"), ("Y", "X", "Y"), ("Y", "Y", "X")]
print(f"Producing random numbers with Mermin Bounds : {get_bounds(uncertainty)}")
nums = generate([0, 1, 0, 1 ,0 ,1 ,0 , 1], 4 , token=None, bases=BASES, uncertainty=uncertainty, check_shots=100, gen_shots=1000)
print(len(nums))
```

```python
filename= "16k_bits"
with open(
        f"tripartite_diqrng/tripartite-diqrng-results/{filename}.txt",
        "a",
    ) as f:
        f.write("".join(nums))
```
