```python
from dotenv import load_dotenv
import time
import os

from tripartite_diqrng.ghz_generate import generate, get_bounds

load_dotenv()
```

# Simulated Backend

```python
uncertainty = 0.5
num_triples = [1,2,4,6]
BASES = [("X", "X", "X"), ("X", "Y", "Y"), ("Y", "X", "Y"), ("Y", "Y", "X")]
print(f"Producing random numbers with Mermin Bounds : {get_bounds(uncertainty)}")
for num in num_triples:
    start = time.time()
    nums = generate([0, 1, 0, 1 ,0 ,1 ,0 , 1], num, token=None, bases=BASES, uncertainty=uncertainty, check_shots=100, gen_shots=1000)
    end = time.time()-start
    print(end)
    print(len(nums))
```
